# From Language to Action: Sequential Decision Making with Vision-Language-Action Models

Sequential decision making and robotics go hand in hand. Modern robotics still relies on Reinforcement Learning (RL) which is used to train policies such that when provided with the state, the next action in the sequence is output. The real-world is full of complexity though, and the sheer diversity of real-world tasks has proved particularly challenging for robots to adapt to when using RL methods. There is a lot of demand for more intuitive, human-compatible interfaces in robotics. This has led to the emergence of Vision-Language-Action Models (VLAs), which attempt to bridge the gap between human instruction and robotic execution. This post describes the traditional RL framework, explores the architecture and promise of VLAs, and examines emerging hybrid models.

## Traditional Sequential Decision Making (Classical RL Pipelines)

In the classical RL formulation, an agent learns apolicy $\pi(a|s)$ that maps a perceivedstate $s$ to an optimalaction $a$. This learning process is governed by areward $R$ that quantifies the desirability of the agent’s actions in the environment.

### Limitations and Challenges

Classical approaches have limitations:

1. Reward Engineering: manuall defined, dense reward functions are often necessary to make this work. It requires a lot of effort to manually tune the rewards to encapsulate the desired behavior. In robotics, sparse rewards can be common when the task reward is binary (e.g. success/failure) which can make effective learning difficult.

2. State Representation/Observability: classical RL relies on implicivtly encoding the goal into the reward function which makes generalization to new tasks non-trivial and can require retraining for a new objective. Because the policy is encoded in the rewards, it is not easy to observe how the state and actions result in the robots behavior in a human-like way.

3. Needs Lots of Data: In order to train the policy, the robot must have many interactions with the environment, building a very large dataset of experiences. Collecting real-world data in robotics is expensive and difficult. There can also be safety concerns with trying to get representative data for all possible environment interactions.


## Language-Grounded Control (Vision-Language-Action Models (VLAs))

Vision-Language-Action Models (VLAs) are new on the robotics scene. VLAs are seen as a a bridge between traditional sequential decision making and high-level human understanding. At their core, VLAs are large-scale neural networks, often based on the Transformer architecture. Most VLAs are built upon pretrained models that have been trained on massive datasets.

Instead of learning from a specifically designed reward, VLAs get their "goals" from natural language instructions. In a robotics context, imagine telling a household robot to "take out the trash".

### Vision-Language Pretraining
VLAs can leverage pre-trained large-scale models (e.g. CLIP, ViT) to create shared embeddings for visual observations and language instructions. This enables the models to inherently understand concepts like object identity, color, size, spatial relationship, etc. before doing any action-specific training.

### Action Prediction as Sequence Modeling
The decision-making process is often framed as a sequence prediction task. Given a history of observations, the current instruction, and the previous actions, the model predicts the next action primitive (or token). This allows the model to benefit from established capabilities in Large Language Models (LLMs) for structured sequence generation.

### Zero-Shot and Few-Shot Generalization
One of the most promising features of VLAs is their ability to generalize, potentially in a way that a classical RL policy cannot. By conditioning the policy on a language instruciton, the model can infer the required sequence of steps by leveraging its world knowledge and the undersanding encoded into the model during the large-scale pretraining. In summary, VLAs are promising for simplifying the human-robot interface because they may allow us to bypass the need for explicit reward engineering by treating the language isntruction as the goal specification. This can allow for a more understandable reward structure for a human observer, but also could provide better generalization for embodied agents.



## Hybrid Architectures (Combining RL + VLAs)

While VLAs represent an ecxiting architecture for robotics, there are limitations. VLAs that can produce low-level control behaviors (e.g. joint actuation) are only just being explored and it's not clear yet whether precise, low-level control from a VLA will be usable and safe for robotic systems. Fortunately, RL has been shown to work very well for low-level control. That gives us the opportunity to combine the language grounded benefits of VLAs with precise low-level control of RL policies to development hybrid approaches for mobile robots!

### Hierarchical VL-RL Frameworks

1. VLA Planner: The VLA component serves as the high-level planner or task decomposition engine. Then given the global language instruction (e.g., "make coffee"), the VLA generates a sequence of sub-goals/sub-tasks in natural language or a latent sub-goal space (e.g., "put the cup under the spout," "press the brew button").
2. RL Low-Level Controller: A classical RL agent (or a library of pre-trained skill policies) is then responsible for low-level, continuous control to execute the current sub-goal. This low-level policy operates in a localized observation space and can be trained with a dense, engineered reward function specific to that sub-task (e.g., a reward for minimizing the distance to the target cup location).

This decoupling allows for the learning of a set of reusable primitive skills (e.g., 'grasp', 'push', 'move to'). The VLA's planning capability then becomes one of recomposing these primitive skills in novel ways to solve new, complex, long-horizon tasks, achieving both generality from the VLA and the low-level precision control from the RL controller.

### Limitations and Challenges of VLAs

1. Scaling & Data: a significant challenge to VLAs is the cost and logistical difficult of acquiring high-quality diverse, multi-modal robotic data.
2. Generalization for Out-of-Distribution Tasks: VLAs can still see many failure modes when attempting to generalize, especially if desired items or objects are not in the image frame.
3. Safety & Explainability: robots operating in proximity of humans will need to be provably safe and demonstrate explainable and predictable behaviors.
