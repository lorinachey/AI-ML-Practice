### Loss Functions
These measure how right or wrong the model is.

* nn.MSELoss - measures average squared difference (common for regression tasks like predicting numbers)
* nn.CrossEntropyLoss - common for multi-class classification (picking 1 out of many categories)
* nn.L1Loss - measures average absolute difference (regression that is less sensitive to outliers)
* nn.BCEWithLogitsLoss - for binary classification (yes/no, t/f decisions)
* nn.NLLLoss - multiclass classification when using logsoftmax activation
* nn.SmoothL1Loss - regression that balances MSE and L1Loss benefits (also called Huber loss)
* nn.KLDivLoss - measures difference between two probability distributions

### Backward
Diagnoses how each parameter contributes to the error.

* Gradients can be thought of as diagnostic scores that tell you what contributed to the loss and in what direction
  * Posiive values tell you that increasing this weight will make the loss worse
  * Negative values tell you that increasing this weight will reduce the loss
* Backward only calculates the gradients, the updates happen later when you call step

### Optimizer Step
Optimizers use the diagnostic scores from backward to update the model weights.

* SGD is a simple and commonly used aproach.
  * Learning rate scales the gradient, but too small of a rate and it takes forever. Too large of a rate and you can miss the minimum.
* Adam is an optimizaer that can adapt to each weight individually.
  * It can figure out which weights need big adjustments and which ones need fine-tuning.
* Remember that you need to call "zero_grad" every itertion before calling backward if you don't want the gradient scores to accumulate.
